```python
import numpy as np
import pandas as pd


dataframe1 = pd.DataFrame({
    "ProductID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "NameofProd": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer",
                   "Scanner", "Webcam", "Headset", "Router", "Tablet"],
    "Price": [55000, 500, 1200, 9000, 7500, 6500, 2500, 1800, 3200, 21000],
})

dataframe2 = pd.DataFrame({
    "ProductID": [101, 103, 105, 107, 109, 111, 112, 113, 114, 115],
    "Company_Location": ["Chennai", "Bengaluru", "Mumbai", "Delhi", "Hyderabad",
                         "Pune", "Kolkata", "Coimbatore", "Jaipur", "Noida"],
})

dataframe3 = pd.DataFrame({
    "ProductID": [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],
    "NameofProd": ["SSD", "HDD", "RAM", "Pen Drive", "Speaker",
                   "Charger", "Power Bank", "Smart Watch", "Earbuds", "Projector"],
    "Price": [4500, 3800, 3000, 600, 2200, 900, 1500, 7000, 2000, 28000],
})

print("dataframe1:\n", dataframe1, "\n")
print("dataframe2:\n", dataframe2, "\n")
print("dataframe3:\n", dataframe3, "\n")
```

```python
# 1. Append records of dataframe1 and dataframe3 (rbind in R)
appended = pd.concat([dataframe1, dataframe3], ignore_index=True)
print("1. Append dataframe1 + dataframe3 (rows):\n", appended, "\n")
```

```python
# 2. Concatenate wherever applicable
row_concat = pd.concat([dataframe1, dataframe3], axis=0, ignore_index=True)
col_concat = pd.concat([dataframe1, dataframe2], axis=1)
print("2a. Row-wise concatenation (df1, df3):\n", row_concat, "\n")
print("2b. Column-wise concatenation (df1, df2):\n", col_concat, "\n")
```

```python
# 3. Merge dataframe1 and dataframe2 using ProductID
merged = pd.merge(dataframe1, dataframe2, on="ProductID")
print("3. Merge on ProductID:\n", merged, "\n")
```

```python
# 4. Join operations between dataframe1 and dataframe2
inner = pd.merge(dataframe1, dataframe2, on="ProductID", how="inner")
left = pd.merge(dataframe1, dataframe2, on="ProductID", how="left")
right = pd.merge(dataframe1, dataframe2, on="ProductID", how="right")
outer = pd.merge(dataframe1, dataframe2, on="ProductID", how="outer")
print("4a. Inner Join:\n", inner, "\n")
print("4b. Left Join:\n", left, "\n")
print("4c. Right Join:\n", right, "\n")
print("4d. Full Outer Join:\n", outer, "\n")
```

```python
#DATA TRANSFORMATIONS
REG_LAST4 = 2710
Balaji = np.array([REG_LAST4, 45, 78, 12, 90, 33, 67, 21, 56, 84], dtype=float)
print("Vector (Balaji):", Balaji, "\n")

x = Balaji
xmin, xmax = x.min(), x.max()


def rescale(x, new_min, new_max):
    """Min-max normalisation to an arbitrary range [new_min, new_max]."""
    return (x - x.min()) / (x.max() - x.min()) * (new_max - new_min) + new_min
```

```python
# 1. Normalisation to ranges
norm_m3_p3 = rescale(x, -3, 3)
norm_m1_p1 = rescale(x, -1, 1)
norm_0_1 = rescale(x, 0, 1)          # Min-Max normalisation
print("Normalised to [-3, +3]:", np.round(norm_m3_p3, 4))
print("Normalised to [-1, +1]:", np.round(norm_m1_p1, 4))
print("Min-Max  [0, 1]       :", np.round(norm_0_1, 4), "\n")
```

```python
# 2. Z-score (sample standard deviation, same as R's sd())
z_score = (x - x.mean()) / x.std(ddof=1)
print("Z-score:", np.round(z_score, 4), "\n")
```

```python
# 3. Decimal scaling: x / 10^j where j is the smallest integer with max(|x'|) < 1
j = int(np.floor(np.log10(np.abs(x).max()))) + 1
decimal_scaled = x / (10 ** j)
print(f"Decimal scaling (j = {j}):", np.round(decimal_scaled, 4), "\n")

summary = pd.DataFrame({
    "Original": x,
    "[-3,+3]": norm_m3_p3,
    "[-1,+1]": norm_m1_p1,
    "[0,1]": norm_0_1,
    "Z-score": z_score,
    "DecimalScaling": decimal_scaled,
}).round(4)
print(summary)
```
