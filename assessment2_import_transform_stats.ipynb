```python
import numpy as np
import pandas as pd

CSV_PATH = "orange_juice_withmissing.csv"
```

```python
# 1. Import dataset
df = pd.read_csv(CSV_PATH)
print("1. Dataset imported.\n")
```

```python
# 2. First 10 records
print("2. First 10 records:\n", df.head(10), "\n")
```

```python
# 3. Last 10 records
print("3. Last 10 records:\n", df.tail(10), "\n")
```

```python
# 4. Number of rows and columns
print(f"4. Rows: {df.shape[0]}, Columns: {df.shape[1]}\n")
```

```python
# 5. Column names
print("5. Variables:", list(df.columns), "\n")
```

```python
# 6. Data types / structure
print("6. Structure:")
df.info()
print()
```

```python
# 7. Summary statistics
print("7. Summary statistics:\n", df.describe(include="all").T, "\n")
```

```python
# 8-10. Missing values
print("8. Missing values in each column:\n", df.isna().sum(), "\n")
print("9. Total missing values in dataset:", int(df.isna().sum().sum()), "\n")
print("10. Number of missing values per column:\n", df.isna().sum().to_dict(), "\n")
```

```python
# 11. Rows with at least one missing value
rows_with_na = df[df.isna().any(axis=1)]
print(f"11. Rows with >=1 missing value: {len(rows_with_na)}")
print(rows_with_na.head(10), "\n")
```

```python
# 12. Remove all rows containing missing values
df_clean = df.dropna()
print(f"12. Cleaned dataset shape: {df_clean.shape}")
print(df_clean.head(10), "\n")
```

```python
# Work on a copy for imputation
d = df.copy()
```

```python
# 13. Mean imputation (PriceMM)
d["PriceMM"] = d["PriceMM"].fillna(d["PriceMM"].mean())
print("13. PriceMM missing after mean imputation:", d["PriceMM"].isna().sum())
```

```python
# 14. Median imputation (LoyalCH)
d["LoyalCH"] = d["LoyalCH"].fillna(d["LoyalCH"].median())
print("14. LoyalCH missing after median imputation:", d["LoyalCH"].isna().sum())
```

```python
# 15. Mode imputation (categorical column). Store7 / Purchase are categorical.
d["Store7"] = d["Store7"].fillna(d["Store7"].mode()[0])
print("15. Store7 missing after mode imputation:", d["Store7"].isna().sum(), "\n")
```

```python
# 16. Forward fill (equivalent of zoo::na.locf in R)
ffill_df = df.copy().ffill()
print("16. Forward fill - remaining NaN per column:\n", ffill_df.isna().sum()[lambda s: s > 0], "\n")
```

```python
# 17. Backward fill (na.locf with fromLast = TRUE)
bfill_df = df.copy().bfill()
print("17. Backward fill - remaining NaN per column:\n", bfill_df.isna().sum()[lambda s: s > 0], "\n")
```

```python
# 18. Replace remaining missing values with a constant (0 for numeric, 'Unknown' for text)
d = d.ffill().bfill()  # first fill using locf in both directions
num_cols = d.select_dtypes(include=np.number).columns
cat_cols = d.select_dtypes(exclude=np.number).columns
d[num_cols] = d[num_cols].fillna(0)
d[cat_cols] = d[cat_cols].fillna("Unknown")
print("18. Total missing after constant fill:", int(d.isna().sum().sum()), "\n")
```

```python
# 19. Remove duplicate records
print("19. Duplicate rows found:", int(d.duplicated().sum()))
d = d.drop_duplicates().reset_index(drop=True)
print("    Shape after removing duplicates:", d.shape, "\n")
```

```python
# 20. Convert appropriate variables into categories (factors)
for c in ["Purchase", "Store7", "StoreID", "STORE", "SpecialCH", "SpecialMM"]:
    d[c] = d[c].astype("category")
print("20. Data types after converting to category:\n", d.dtypes, "\n")
```

```python
# 21. Rename columns
d = d.rename(columns={"WeekofPurchase": "Week", "LoyalCH": "LoyaltyCH"})
print("21. Renamed columns:", list(d.columns), "\n")
```

```python
# 22. Derived column
d["PriceRatio"] = d["PriceMM"] / d["PriceCH"]
print("22. Derived column PriceRatio:\n", d[["PriceCH", "PriceMM", "PriceRatio"]].head(), "\n")
```

```python
# 23. summary()
print("23. Summary statistics:\n", d.describe(include="all").T, "\n")
```

```python
# 24. Min and max
num = d.select_dtypes(include=np.number)
print("24. Minimum values:\n", num.min(), "\n")
print("    Maximum values:\n", num.max(), "\n")
```

```python
# 25. Range of a numeric variable
col = "PriceMM"
print(f"25. Range of {col}: {d[col].max() - d[col].min():.4f}\n")
```

```python
# 26. Variance
print(f"26. Variance of {col}: {d[col].var():.6f}\n")
```

```python
# 27. Standard deviation
print(f"27. Standard deviation of {col}: {d[col].std():.6f}\n")
```

```python
# 28. IQR
q1, q3 = d[col].quantile([0.25, 0.75])
print(f"28. IQR of {col}: {q3 - q1:.4f}  (Q1={q1:.4f}, Q3={q3:.4f})")
```
